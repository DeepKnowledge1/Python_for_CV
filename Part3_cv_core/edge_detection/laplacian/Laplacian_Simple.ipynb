
# Sobel Edge Detection (Simple)

This minimal notebook loads (or generates) a grayscale image and applies **Sobel X**, **Sobel Y**, and a **combined** edge map using OpenCV.


```python

import os
import numpy as np
import cv2
import matplotlib.pyplot as plt

print("OpenCV:", cv2.__version__)
```

```python

# Set this to your image file name if you have one in the same folder as the notebook
IMG_PATH = "example.jpg"

def make_synthetic(w=384, h=256):
    img = np.zeros((h, w), dtype=np.uint8)
    cv2.rectangle(img, (20, 20), (160, 120), 200, -1)
    cv2.circle(img, (260, 160), 45, 150, 2)
    cv2.line(img, (10, h-10), (w-10, 10), 255, 2)
    cv2.putText(img, "Sobel", (120, 80), cv2.FONT_HERSHEY_SIMPLEX, 1, 220, 2, cv2.LINE_AA)
    return img

if os.path.exists(IMG_PATH):
    img = cv2.imread(IMG_PATH, cv2.IMREAD_GRAYSCALE)
    if img is None:
        img = make_synthetic()
else:
    img = make_synthetic()

print("Image shape:", img.shape)
plt.figure(figsize=(5,3))
plt.imshow(img, cmap='gray')
plt.title('Original')
plt.axis('off')
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Step 1: Create a simple synthetic image
# img = np.array([
#     [10, 10, 10, 10, 10],
#     [10, 50, 50, 50, 10],
#     [10, 50, 200, 50, 10],
#     [10, 50, 50, 50, 10],
#     [10, 10, 10, 10, 10]
# ], dtype=np.float32)

# Step 2: Define Laplacian kernel (4-neighbor)
kernel = np.array([[0, 1, 0],
                   [1, -4, 1],
                   [0, 1, 0]], dtype=np.float32)

# Step 3: Apply convolution
laplacian = cv2.filter2D(img, -1, kernel)

# Step 4: Normalize for display
laplacian_abs = np.abs(laplacian)
laplacian_norm = (laplacian_abs / laplacian_abs.max()) * 255
laplacian_norm = laplacian_norm.astype(np.uint8)

# Step 5: Visualization
fig, axs = plt.subplots(1, 3, figsize=(10, 3))
axs[0].imshow(img, cmap='gray')
axs[0].set_title("Original Image")
axs[1].imshow(laplacian, cmap='gray')
axs[1].set_title("Laplacian (raw)")
axs[2].imshow(laplacian_norm, cmap='gray')
axs[2].set_title("Laplacian (normalized)")

for ax in axs:
    ax.axis('off')
plt.show()
```
