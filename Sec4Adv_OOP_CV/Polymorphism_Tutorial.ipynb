## 🧠 1. What is Polymorphism?
**Polymorphism** means “**many forms**.” In Python, it allows the same method or function name to behave differently based on the object calling it.

👉 It’s a way to use **one interface** for **multiple implementations**.

### 🔹 Real Life Analogy:
Let’s say you press the **start** button:
- On a **TV**, it turns it on.
- On a **Washing Machine**, it begins a wash cycle.
- On a **Microwave**, it starts heating.

Same **action**, but **different behavior** based on the object.

### Key Benefits of Polymorphism

1. **Code Reusability**: Polymorphism promotes code reuse by allowing the same interface to be used with different implementations.
2. **Flexibility**: It makes code more flexible and adaptable to changes.
3. **Extensibility**: New classes can be added with minimal changes to existing code.
4. **Simplicity**: It simplifies code by allowing the same method name to be used for different operations.
5. **Abstraction**: It supports abstraction by focusing on what an object does rather than how it does it.


## 🧪 2. Polymorphism Using Functions

```python
class Dog:
    def speak(self):
        return "Woof!"

class Cat:
    def speak(self):
        return "Meow!"
    
def animal_sound(animal):
    print(animal.speak())

dog = Dog()
cat = Cat()

animal_sound(dog)  # Woof!
animal_sound(cat)  # Meow!
```

## 🔁 3. Polymorphism in Methods

```python
class Bird:
    def fly(self):
        print("Bird is flying")

class Eagle(Bird):
    pass
    def fly(self):
        print("Eagle soars high")

class Penguin(Bird):
    def fly(self):
        print("Penguin can't fly 😅")

birds = [Bird(), Eagle(), Penguin()]

for b in birds:
    b.fly()
```

## 🧰 4. Polymorphism with Abstract Base Classes

```python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

class Circle(Shape):
    def area(self):
        return 3.14 * 5 * 5
# 
class Square(Shape):
    def area(self):
        return 4 * 4

shapes = [Circle(), Square()]
for s in shapes:
    print(s.area())
```

## 📦 5. Real-World Example: File Readers

```python
class PDFReader:
    def read(self):
        return "Reading a PDF file"

class WordReader:
    def read(self):
        return "Reading a Word document"

class ExcelReader:
    def read(self):
        return "Reading an Excel sheet"

def read_file(reader):
    print(reader.read())

read_file(PDFReader())
read_file(WordReader())
read_file(ExcelReader())
```
