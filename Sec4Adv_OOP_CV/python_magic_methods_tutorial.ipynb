## 1. What are Magic Methods?

### 🪄 The Magic Behind Python

Magic methods (also called **dunder methods** because they have **d**ouble **under**scores) are special methods in Python that allow you to define how your objects behave with built-in operations.

**Think of it this way:** When you write `5 + 3`, Python is actually calling `5.__add__(3)` behind the scenes!

### Why are they called "Dunder" methods?
- **D**ouble **Under**score methods
- They start and end with `__` (double underscores)
- Examples: `__init__`, `__str__`, `__add__`

Let's see this in action:

```python
# Let's see what happens when we add numbers
result1 = 5 + 3
result2 = (5).__add__(3)  # This is what Python does internally!
result3 = int.__add__(5,3)

print(f"5 + 3 = {result1}")
print(f"(5).__add__(3) = {result2}")
print(f"int.__add__(3,5) = ", result3)
print(f"Are they the same? {result1 == result2== result3}")
```

## 4. Arithmetic Operations

### 🧮 Making Objects Do Math!

Arithmetic magic methods let your objects work with mathematical operators like `+`, `-`, `*`, `/`.

**Common Arithmetic Methods:**
- `__add__(self, other)` → `+` (addition)
- `__sub__(self, other)` → `-` (subtraction)
- `__mul__(self, other)` → `*` (multiplication)
- `__truediv__(self, other)` → `/` (division)
- `__pow__(self, other)` → `**` (power)

### Let's Build a Vector Class!

```python
import math

class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    
    def __str__(self):
        return f"Vector({self.x}, {self.y})"
    
    def __repr__(self):
        return f"Vector(x={self.x}, y={self.y})"
    
    def __add__(self, other):
        """Add two vectors"""
        if not isinstance(other, Vector):
            return NotImplemented
        return Vector(self.x + other.x, self.y + other.y)
    
    def __sub__(self, other):
        """Subtract two vectors"""
        if not isinstance(other, Vector):
            return NotImplemented
        return Vector(self.x - other.x, self.y - other.y)
    
    def __mul__(self, scalar):
        """Multiply vector by a scalar"""
        if not isinstance(scalar, (int, float)):
            return NotImplemented
        return Vector(self.x * scalar, self.y * scalar)
    
    def __rmul__(self, scalar):
        """Right multiplication (allows scalar * vector)"""
        return self.__mul__(scalar)
    
    def __eq__(self, other):
        """Check if two vectors are equal"""
        if not isinstance(other, Vector):
            return False
        return self.x == other.x and self.y == other.y
    
    def magnitude(self):
        """Calculate the magnitude of the vector"""
        return math.sqrt(self.x**2 + self.y**2)

# Let's test our Vector class!
v1 = Vector(3, 4)
v2 = Vector(1, 2)

print("Vectors:")
print(f"v1 = {v1}")
print(f"v2 = {v2}")
print()

print("Vector operations:")
print(f"v1 + v2 = {v1 + v2}")
print(f"v1 - v2 = {v1 - v2}")
print(f"v1 * 2 = {v1 * 2}")
print(f"3 * v2 = {3 * v2}")
print()

print("Vector properties:")
print(f"v1 == v2: {v1 == v2}")
print(f"v1 == Vector(3, 4): {v1 == Vector(3, 4)}")
```

## 5. Container Behavior

### 📦 Making Objects Act Like Containers

Container magic methods allow your objects to behave like lists, dictionaries, or other containers.

**Common Container Methods:**
- `__len__(self)` → `len(obj)` (get length)
- `__getitem__(self, key)` → `obj[key]` (get item)
- `__setitem__(self, key, value)` → `obj[key] = value` (set item)
- `__delitem__(self, key)` → `del obj[key]` (delete item)
- `__contains__(self, item)` → `item in obj` (membership test)

### Let's Create a Custom Playlist Class!

```python
class Playlist:
    def __init__(self, name):
        self.name = name
        self.songs = []
    
    def __str__(self):
        return f"Playlist '{self.name}' with {len(self.songs)} songs"
    
    def __len__(self):
        """Return the number of songs in the playlist"""
        return len(self.songs)
    
    def __getitem__(self, index):
        """Get a song by index"""
        return self.songs[index]
    
    def __setitem__(self, index, song):
        """Set a song at a specific index"""
        self.songs[index] = song
    
    def __delitem__(self, index):
        """Delete a song by index"""
        del self.songs[index]
    
    def __contains__(self, song):
        """Check if a song is in the playlist"""
        return song in self.songs
    
    def add_song(self, song):
        """Add a song to the playlist"""
        self.songs.append(song)
    
    def __iter__(self):
        """Make the playlist iterable"""
        return iter(self.songs)

# Let's test our Playlist!
my_playlist = Playlist("My Favorites")

# Add some songs
my_playlist.add_song("Bohemian Rhapsody")
my_playlist.add_song("Stairway to Heaven")
my_playlist.add_song("Hotel California")

print(f"Playlist: {my_playlist}")
print(f"Length: {len(my_playlist)}")
print()

# Access songs by index
print("Songs by index:")
print(f"First song: {my_playlist[0]}")
print(f"Last song: {my_playlist[-1]}")
print()

# Check if song is in playlist
print("Membership tests:")
print(f"'Bohemian Rhapsody' in playlist: {'Bohemian Rhapsody' in my_playlist}")
print(f"'Yesterday' in playlist: {'Yesterday' in my_playlist}")
print()

# Iterate through playlist
print("All songs:")
for i, song in enumerate(my_playlist, 1):
    print(f"{i}. {song}")
```

## 6. Callable Objects

### 📞 Making Objects Behave Like Functions

The `__call__` method allows your objects to be called like functions!

**Syntax:** `obj()` calls `obj.__call__()`

### Let's Create a Counter Class!

```python
class Counter:
    def __init__(self, start=0, step=1):
        self.value = start
        self.step = step
    
    def __str__(self):
        return f"Counter(value={self.value}, step={self.step})"
    
    def __call__(self):
        """When called, increment and return the current value"""
        current = self.value
        self.value += self.step
        return current
    
    def reset(self, start=0):
        """Reset the counter"""
        self.value = start

# Let's test our callable Counter!
counter = Counter(start=10, step=2)
print(f"Initial counter: {counter}")
print()

print("Calling the counter:")
print(f"counter() = {counter()}")
print(f"counter() = {counter()}")
print(f"counter() = {counter()}")
print(f"Current state: {counter}")
```

## 7. Tips & Best Practices

### 🎯 Writing Better Magic Methods

Here are essential tips for using magic methods effectively:

#### 1. **Always Check Types**

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    
    def __add__(self, other):
        # ✅ Good: Check the type first
        if not isinstance(other, Point):
            return NotImplemented
        return Point(self.x + other.x, self.y + other.y)
    
    def __str__(self):
        return f"Point({self.x}, {self.y})"

# This works correctly
p1 = Point(1, 2)
p2 = Point(3, 4)
print(f"p1 + p2 = {p1 + p2}")

# This fails gracefully
try:
    result = p1 + "invalid"
    print(f"p1 + 'invalid' = {result}")
except TypeError as e:
    print(f"Error (as expected): {e}")
```

#### 2. **Return `NotImplemented`, Not `False`**

```python
class Temperature:
    def __init__(self, celsius):
        self.celsius = celsius
    
    def __eq__(self, other):
        if not isinstance(other, Temperature):
            # ✅ Good: Return NotImplemented
            return NotImplemented
        return self.celsius == other.celsius
    
    def __str__(self):
        return f"{self.celsius}°C"

temp1 = Temperature(25)
temp2 = Temperature(25)
temp3 = Temperature(30)

print(f"temp1 == temp2: {temp1 == temp2}")
print(f"temp1 == temp3: {temp1 == temp3}")
print(f"temp1 == 'hot': {temp1 == 'hot'}")
```

## 8. Common Mistakes to Avoid

### ❌ Mistake #1: Modifying Objects In-Place

```python
class BadVector:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    
    def __add__(self, other):
        # ❌ Bad: Modifies the original object
        self.x += other.x
        self.y += other.y
        return self
    
    def __str__(self):
        return f"Vector({self.x}, {self.y})"

class GoodVector:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    
    def __add__(self, other):
        # ✅ Good: Returns a new object
        return GoodVector(self.x + other.x, self.y + other.y)
    
    def __str__(self):
        return f"Vector({self.x}, {self.y})"

# Demonstrate the problem
print("❌ Bad example:")
bad_v1 = BadVector(1, 2)
bad_v2 = BadVector(3, 4)
print(f"Before: bad_v1 = {bad_v1}")
bad_result = bad_v1 + bad_v2
print(f"After: bad_v1 = {bad_v1}")
print(f"bad_v1 is bad_result: {bad_v1 is bad_result}")
print()

print("✅ Good example:")
good_v1 = GoodVector(1, 2)
good_v2 = GoodVector(3, 4)
print(f"Before: good_v1 = {good_v1}")
good_result = good_v1 + good_v2
print(f"After: good_v1 = {good_v1}")
print(f"good_v1 is good_result: {good_v1 is good_result}")
```

## 9. Practice Exercises

### 🎯 Exercise 1: Complete Bank Account Class

Create a `BankAccount` class with magic methods:

```python
class BankAccount:
    def __init__(self, account_number, initial_balance=0):
        self.account_number = account_number
        self.balance = initial_balance
    
    def __str__(self):
        return f"Account {self.account_number}: ${self.balance:.2f}"
    
    def __repr__(self):
        return f"BankAccount('{self.account_number}', {self.balance})"
    
    def __eq__(self, other):
        if not isinstance(other, BankAccount):
            return False
        return self.account_number == other.account_number
    
    def __add__(self, amount):
        if not isinstance(amount, (int, float)):
            return NotImplemented
        return BankAccount(self.account_number, self.balance + amount)
    
    def __sub__(self, amount):
        if not isinstance(amount, (int, float)):
            return NotImplemented
        new_balance = self.balance - amount
        if new_balance < 0:
            raise ValueError("Insufficient funds")
        return BankAccount(self.account_number, new_balance)
    
    def __lt__(self, other):
        if not isinstance(other, BankAccount):
            return NotImplemented
        return self.balance < other.balance

# Test the BankAccount
account1 = BankAccount("12345", 1000)
account2 = BankAccount("67890", 500)
account3 = BankAccount("12345", 2000)

print(f"Account 1: {account1}")
print(f"Account 2: {account2}")
print(f"account1 == account3: {account1 == account3}")
print(f"account1 < account2: {account1 < account2}")
print(f"account1 + 500: {account1 + 500}")
print(f"account1 - 200: {account1 - 200}")
```
