## 🧠 1. What is Encapsulation?

**Encapsulation** is the concept of **hiding internal object details** and providing a **clean interface** to interact with them.

📌 **Goal:** Prevent external access to sensitive parts of your class and provide controlled access through methods.

### 💬 Real-Life Analogy
Think of a **vending machine**:
- You don’t open the machine and grab a soda (internal/private access).
- You press a button (clean method), and the soda comes out (controlled output).

### Access Level Summary

| Convention | Example | Meaning | Accessibility |
|------------|---------|---------|---------------|
| `name` | `self.account_number` | Public | Accessible everywhere |
| `_name` | `self._balance` | Protected | Internal use, subclasses |
| `__name` | `self.__pin` | Private | Very internal, name mangled |

## 🧪 2. Basic Class Without Encapsulation (Bad Practice)

````python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

person = Person("Ali", 20)
person.age = -5  # This should not be allowed!
print(person.age)
````

## ✅ 3. Encapsulation Using Getters and Setters (Old Style)

````python
class Person:
    def __init__(self, name, age):
        self._name = name
        self._age = age

    def get_age(self):
        return self._age

    def set_age(self, value):
        if value < 0:
            raise ValueError("Age cannot be negative")
        self._age = value

p = Person("Ali", 20)
print(p.get_age())     # 20
p.set_age(25)
print(p.get_age())     # 25
````

## ✅ 4. Pythonic Way: `@property` Decorator (Modern & Clean)

````python
class Person:
    def __init__(self, name, age):
        self._name = name
        self._age = age

    @property
    def age(self):
        return self._age

    @age.setter
    def age(self, value):
        if value < 0:
            raise ValueError("Age cannot be negative")
        self._age = value

p = Person("Ali", 20)
print(p.age)
p.age = 25
print(p.age)
````

## 🔍 5. Read-Only Property

````python
class Circle:
    def __init__(self, radius):
        self._radius = radius

    @property
    def radius(self):
        return self._radius

    @property
    def area(self):
        return 3.14 * (self._radius ** 2)

c = Circle(5)
print(c.area)
# c.radius = 100  # ❌ Error – read-only
````

## 📦 6. Why Use Encapsulation?

- ✅ Protect internal state
- ✅ Validate input before setting values
- ✅ Control access
- ✅ Read-only or write-only logic
- ✅ Maintainability & Clean API

## ⚠️ 7. Common Mistakes

❌ Forgetting to prefix attributes with `_`  
❌ Defining a setter without a getter  
❌ Returning `self.age` inside `@property` (infinite recursion)  
❌ Allowing invalid values (negative age, zero salary, etc.)

## 👨‍💻 8. Real-World Example

````python
import numpy as np
from datetime import datetime
import hashlib

class ImageProcessor:
    """A class for handling image data with proper encapsulation"""
    
    def __init__(self, data, name="Untitled"):
        # Private attributes
        self._data = None
        self._name = ""
        self._created_at = datetime.now()
        self._modified_at = None
        
        # Use setters for validation
        self.name = name
        self.data = data
    
    @property
    def name(self):
        """Get the image name"""
        return self._name
    
    @name.setter
    def name(self, value):
        """Set the image name with validation"""
        if not isinstance(value, str):
            raise TypeError("Name must be a string")
        if not value.strip():
            raise ValueError("Name cannot be empty")
        self._name = value.strip()
        self._update_modified_time()
    
    @property
    def data(self):
        """Get the image data"""
        return self._data
    
    @data.setter
    def data(self, value):
        """Set the image data with validation"""
        if not isinstance(value, np.ndarray):
            raise TypeError("Data must be a NumPy array")
        
        if value.ndim not in [2, 3]:
            raise ValueError("Data must be 2D (grayscale) or 3D (color) array")
        
        if value.ndim == 3 and value.shape[2] not in [1, 3, 4]:
            raise ValueError("Color images must have 1, 3, or 4 channels")
        
        # Store a copy to prevent external modification
        self._data = value.copy()
        self._update_modified_time()
    
    @property
    def width(self):
        """Image width in pixels (read-only)"""
        return self._data.shape[1] if self._data is not None else 0
    
    @property
    def height(self):
        """Image height in pixels (read-only)"""
        return self._data.shape[0] if self._data is not None else 0
    
    @property
    def channels(self):
        """Number of color channels (read-only)"""
        if self._data is None:
            return 0
        return 1 if self._data.ndim == 2 else self._data.shape[2]
    
    @property
    def shape(self):
        """Image shape tuple (read-only)"""
        return self._data.shape if self._data is not None else (0, 0)
    
    @property
    def size(self):
        """Total number of pixels (read-only)"""
        return self.width * self.height
    
    @property
    def aspect_ratio(self):
        """Width to height ratio (read-only)"""
        return self.width / self.height if self.height > 0 else 0
    
    @property
    def is_grayscale(self):
        """Check if image is grayscale (read-only)"""
        return self.channels == 1
    
    @property
    def is_color(self):
        """Check if image is color (read-only)"""
        return self.channels >= 3
    
    @property
    def has_alpha(self):
        """Check if image has alpha channel (read-only)"""
        return self.channels == 4
    
    @property
    def data_type(self):
        """NumPy data type of the image (read-only)"""
        return self._data.dtype if self._data is not None else None
    
    @property
    def memory_usage(self):
        """Memory usage in bytes (read-only)"""
        return self._data.nbytes if self._data is not None else 0
    
    @property
    def checksum(self):
        """MD5 checksum of image data (read-only)"""
        if self._data is None:
            return None
        return hashlib.md5(self._data.tobytes()).hexdigest()
    
    @property
    def created_at(self):
        """Creation timestamp (read-only)"""
        return self._created_at
v 
    @property
    def modified_at(self):
        """Last modification timestamp (read-only)"""
        return self._modified_at
    
    def _update_modified_time(self):
        """Internal method to update modification time"""
        self._modified_at = datetime.now()
    
    def get_stats(self):
        """Get comprehensive image statistics"""
        if self._data is None:
            return "No image data"
        
        stats = {
            'name': self.name,
            'dimensions': f"{self.width}x{self.height}",
            'channels': self.channels,
            'type': 'Color' if self.is_color else 'Grayscale',
            'aspect_ratio': f"{self.aspect_ratio:.2f}",
            'data_type': str(self.data_type),
            'memory_usage': f"{self.memory_usage / 1024:.1f} KB",
            'checksum': self.checksum[:8] + '...',
            'created': self.created_at.strftime('%Y-%m-%d %H:%M:%S'),
            'modified': self.modified_at.strftime('%Y-%m-%d %H:%M:%S') if self.modified_at else 'Never'
        }
        return stats
    
    def __str__(self):
        return f"ImageProcessor('{self.name}', {self.width}x{self.height}, {self.channels} channels)"
    
    def __repr__(self):
        return self.__str__()

# Create sample images for testing
print("Creating sample images...\n")

# Grayscale image
gray_data = np.random.randint(0, 256, (200, 300), dtype=np.uint8)
gray_img = ImageProcessor(gray_data, "Sample Grayscale")

# Color image
color_data = np.random.randint(0, 256, (200, 300, 3), dtype=np.uint8)
color_img = ImageProcessor(color_data, "Sample Color")

# RGBA image
rgba_data = np.random.randint(0, 256, (150, 200, 4), dtype=np.uint8)
rgba_img = ImageProcessor(rgba_data, "Sample RGBA")

# Display information
for img in [gray_img, color_img, rgba_img]:
    print(f"=== {img.name} ===")
    stats = img.get_stats()
    for key, value in stats.items():
        print(f"{key.replace('_', ' ').title()}: {value}")
    print()

# Demonstrate property usage
print("=== Property Demonstrations ===")
print(f"Color image aspect ratio: {color_img.aspect_ratio:.2f}")
print(f"Is grayscale: {color_img.is_grayscale}")
print(f"Has alpha channel: {rgba_img.has_alpha}")
print(f"Memory usage: {rgba_img.memory_usage / 1024:.1f} KB")

# Test validation
print("\n=== Validation Tests ===")
try:
    color_img.name = ""  # Empty name
except ValueError as e:
    print(f"❌ Name validation: {e}")

try:
    color_img.data = "invalid data"  # Wrong type
except TypeError as e:
    print(f"❌ Data type validation: {e}")

try:
    invalid_shape = np.random.rand(100, 100, 100, 100)  # 4D array
    color_img.data = invalid_shape
except ValueError as e:
    print(f"❌ Data shape validation: {e}")

# Test read-only properties
try:
    color_img.width = 500  # Try to modify read-only property
except AttributeError as e:
    print(f"❌ Read-only property: {e}")
````

## 9. Summary and Key Takeaways

### What We've Learned

1. **Encapsulation** protects data and provides controlled access through well-defined interfaces
2. **Python access conventions** use naming patterns (`public`, `_protected`, `__private`) to indicate intended usage
3. **Property decorators** provide a Pythonic way to implement getters, setters, and computed properties
4. **Read-only properties** are created by omitting the setter decorator
5. **Validation** can be seamlessly integrated into property setters
6. **Computed properties** calculate values on-demand and can be cached for performance

### When to Use Property Decorators

- **Data validation** is required before storing values
- **Computed values** need to be calculated from other attributes
- **Read-only access** to certain attributes is desired
- **Backward compatibility** when converting simple attributes to methods
- **Logging or debugging** attribute access

### Property Decorators vs. Regular Methods

| Use Properties When | Use Methods When |
|-------------------|------------------|
| Accessing/setting simple values | Performing complex operations |
| Computing derived values | Taking multiple parameters |
| Validating single inputs | Having side effects |
| Maintaining attribute-like interface | Returning different types based on context |

### Final Tips

1. **Start simple** - Begin with basic properties and add complexity as needed
2. **Document behavior** - Clearly document what properties do, especially read-only ones
3. **Test thoroughly** - Validate both normal and edge cases
4. **Consider performance** - Cache expensive computations when appropriate
5. **Be consistent** - Use similar patterns throughout your codebase

Property decorators are a powerful feature that makes Python code more readable, maintainable, and robust. They bridge the gap between simple attribute access and complex method-based operations, providing the best of both worlds.
