# 🐍 Abstract Base Classes in Python: Beginner Tutorial

## 🎯 Goal
Learn how to use `abc.ABC` and `@abstractmethod` to enforce rules in subclasses.

## 📌 What is an Abstract Base Class (ABC)?
An Abstract Base Class is a class that **cannot be instantiated** and is used to define a blueprint for other classes.

## 🔧 Why Use ABCs?
- Enforce consistent APIs
- Provide a common interface
- Prevent incomplete class instantiation

```python
from abc import ABC, abstractmethod
```

## ✨ Creating an Abstract Base Class

```python
class Animal(ABC):
    @abstractmethod
    def speak(self):
        pass
```

## ❌ Cannot Instantiate Abstract Class

```python
# a = Animal()  # Uncomment to see error
```

## ✅ Subclass with Implementation

```python
class Dog(Animal):
    def speak(self):
        return "Woof!"

d = Dog()
print(d.speak())
```

## ❌ Missing Implementation in Subclass

```python
# class Cat(Animal):
#     pass
# c = Cat()  # Uncomment to see error
```

## 🧾 Real-World Analogy
ABC = Blueprint, Concrete class = Real Object

## ✅ Payment System Example

```python
class PaymentMethod(ABC):
    @abstractmethod
    def pay(self, amount):
        pass

class CreditCard(PaymentMethod):
    def pay(self, amount):
        print(f"Paid ${amount} with credit card.")

class PayPal(PaymentMethod):
    def pay(self, amount):
        print(f"Paid ${amount} with PayPal.")
        

p = CreditCard()
p.pay(100)
```

## ⚠️ Multiple Abstract Methods

```python
class Vehicle(ABC):
    @abstractmethod
    def start(self):
        pass

    @abstractmethod
    def stop(self):
        pass
    
    def print():
        print("Welcome ")
    

class Car(Vehicle):
    def start(self):
        pass

    def stop(self):
        pass



c = Car()
```

## 🤯 Abstract Properties

```python
class Shape(ABC):
    @property
    @abstractmethod
    def area(self):
        pass

class Square(Shape):
    def __init__(self, side):
        self._side = side

    @property
    def area(self):
        return self._side ** 2
```

## ❓ FAQ
- ABCs can have regular methods
- Subclasses can be abstract too

## 🔄 ABC vs Interface

| Feature              | Abstract Class | Interface-like |
|---------------------|----------------|----------------|
| Method Types        | Mixed          | All abstract   |
| Use Case            | Shared logic   | Rule enforcing |

## ✅ Summary
