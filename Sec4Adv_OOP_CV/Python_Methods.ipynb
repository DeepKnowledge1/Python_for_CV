# Python Methods: Instance, Class, and Static

## Introduction

In Python, methods are functions that belong to a class. They define the behaviors and actions that objects created from that class can perform. Understanding the different types of methods—instance, class, and static—is crucial for writing effective, organized, and Pythonic code. Each type serves a distinct purpose and interacts with the class and its instances in unique ways.

At its core, object-oriented programming (OOP) in Python revolves around classes and objects. Classes act as blueprints, and objects are instances of those blueprints. Methods are the actions associated with these blueprints or their instances. The key difference between instance, class, and static methods lies in their first argument and, consequently, what they can access and modify.

## 1. Instance Methods

Instance methods are the most common type of methods in Python classes. They operate on an instance of the class and can access and modify the instance's state (i.e., its attributes). When you define an instance method, its first parameter is conventionally named `self`, which refers to the instance of the class itself.

### Definition and Syntax

```python
class MyClass:
    def __init__(self, value):
        self.instance_variable = value

    def instance_method(self, new_value):
        print(f"Current instance variable: {self.instance_variable}")
        self.instance_variable = new_value
        print(f"New instance variable: {self.instance_variable}")

obj = MyClass(10)
obj.instance_method(20)
```

### When to Use Instance Methods

- Access or modify instance variables.
- Method depends on the specific object's state.
- Need to call other instance methods.

**Example Use Case:** A `Car` class with an `accelerate()` method that modifies `speed`.

### Advantages:
- Access to instance state
- Supports encapsulation and polymorphism

### Disadvantages:
- Requires an instance
- Slight overhead from `self`

## 2. Class Methods

Class methods are bound to the class, not an instance. They operate on class variables and are defined with the `@classmethod` decorator. Their first parameter is `cls`, referring to the class itself.

```python
class MyClass:
    class_variable = 0

    def __init__(self, value):
        self.instance_variable = value

    @classmethod
    def class_method(cls, new_value):
        print(f"Current class variable: {cls.class_variable}")
        cls.class_variable = new_value
        print(f"New class variable: {cls.class_variable}")

MyClass.class_method(100)
obj = MyClass(10)
obj.class_method(200)
print(f"Final class variable: {MyClass.class_variable}")
```

### When to Use Class Methods
- Access or modify class-level data
- Create alternative constructors
- Use inheritance-aware logic

**Example Use Case:** `from_string()` constructor in a `Date` class.

### Advantages:
- Access to class state
- Useful for flexible constructors
- Inheritance-aware

### Disadvantages:
- Cannot access instance state
- Potential for misuse in place of static/regular methods

## 3. Static Methods

Static methods do not access class or instance state. They’re defined with the `@staticmethod` decorator and take no implicit first argument.

```python
class MyClass:
    def __init__(self, value):
        self.instance_variable = value

    class_variable = 0

    @staticmethod
    def static_method(x, y):
        return x + y

result1 = MyClass.static_method(5, 3)
print(f"Result via class: {result1}")

obj = MyClass(10)
result2 = obj.static_method(7, 2)
print(f"Result via instance: {result2}")
```

### When to Use Static Methods
- Pure functions that don’t need access to `self` or `cls`
- Utility functions logically related to a class
- Methods that could be standalone, but are placed for organizational reasons

**Example Use Case:** `calculate_area()` method in a `MathOperations` class.

### Advantages:
- No need for `self` or `cls`
- Simple and reusable
- Great for utilities

### Disadvantages:
- Cannot access or modify instance/class state
