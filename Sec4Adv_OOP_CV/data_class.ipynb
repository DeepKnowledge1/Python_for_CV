## 1️⃣ What is a Data Class?

In Python, a **data class** is a simple way to create classes that are **mainly used to store data**, without writing a lot of boilerplate code.

Introduced in **Python 3.7**, data classes automatically generate:
- `__init__()` – constructor
- `__repr__()` – string representation
- `__eq__()` – equality check
- `__hash__()`, `__lt__()`, and more (optional)

## 2️⃣ Classes vs Data Classes

| Feature                | Regular Class                                | Data Class (`@dataclass`)              |
|------------------------|----------------------------------------------|----------------------------------------|
| Code length            | Longer (manual init, repr, eq, etc.)         | Shorter (auto-generated methods)       |
| Boilerplate            | You write everything                         | Python generates for you               |
| Immutability           | Manual with `__setattr__`                    | Easy using `frozen=True`               |
| Default values         | Must define in `__init__()`                  | Set in class body                      |
| Comparison support     | Manual `__eq__()`                            | Automatic                              |

````python
# 3️⃣ Basic Syntax
from dataclasses import dataclass

@dataclass
class Person:
    name: str
    age: int
````

````python
@dataclass
class Book:
    title: str
    author: str = "Unknown"
    pages: int = 0

````

````python
# 5️⃣ Making Data Class Immutable
@dataclass(frozen=True)
class Point:
    x: int
    y: int

p = Point(1, 2)
````

````python
# 6️⃣ Using field() for Mutable Default
from dataclasses import field

@dataclass
class Product:
    name: str
    price: float
    tags: list[str] = field(default_factory=list)

````

## 7️⃣ Advantages of Data Classes ✅
- Less code, cleaner syntax
- Readable and maintainable
- Auto-generated methods
- Built-in comparison and hashing
- `frozen=True` allows immutability
- Supports type hints and default values

## 8️⃣ Disadvantages of Data Classes ❌
- Limited to simple data containers
- Not ideal for behavior-rich objects (use regular classes)
- Slower than plain tuples for very large data sets
- No inheritance of `__post_init__` logic unless explicitly done
- Requires Python 3.7+

## 9️⃣ Interview Questions 💬
1. What is the purpose of the `@dataclass` decorator?
2. How is a data class different from a regular class?
3. When would you use `field(default_factory=...)`?
4. How do you make a data class immutable?
5. Can data classes be inherited?

````python
# 🔟 Tips: __post_init__ customization
@dataclass
class Employee:
    name: str
    salary: float

    def __post_init__(self):
        if self.salary < 0:
            raise ValueError("Salary must be positive")

# e = Employee("Tom", -1000)  # ❌ Raises ValueError
````

````python
# Inheriting data classes
@dataclass
class Animal:
    name: str

@dataclass
class Dog(Animal):
    breed: str

d = Dog("Buddy", "Labrador")
print(d)
````

## 🧪 Quick Quiz
What’s the output?
```python
from dataclasses import dataclass

@dataclass
class A:
    x: int = 10
    y: int = 20

a1 = A()
a2 = A()

print(a1 == a2)
```
**Answer**: `True` (Auto-generated `__eq__` checks values)

## 🧠 Summary
- Use `@dataclass` when your class is mostly data.
- It reduces boilerplate code and adds useful functionality.
- Great for DTOs, configurations, records, and models.
